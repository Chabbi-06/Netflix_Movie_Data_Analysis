 ## Categorizing Vote_Average column
 making 4 categories for Vote_Average
 - popular
 - average
 - below_avg
 - not_popular
uaing categorize_col() function

```python
def categorized_col(df,col,labels):
    edges = [df[col].describe()['min'],
             df[col].describe()['25%'],
             df[col].describe()['50%'],
             df[col].describe()['75%'],
             df[col].describe()['max']]
   
    df[col] = pd.cut(df[col],edges , labels = labels , duplicates = 'drop')
    return df
    
             
```

```python
labels = ['not_popular', 'below-avg' , 'average', 'popular']

categorized_col(df, 'Vote_Average' , labels)
df['Vote_Average'].unique()
```

## Split the genres into a list and then explode our dataframe to have only one genre per row for each movie

```python
df['Genre'] = df['Genre'].str.split(', ')
df= df.explode("Genre").reset_index(drop = True)
df.head()
```

```python
print(df['Genre'].head())
```

```python
# casting column into category

df['Genre'] = df['Genre'].astype('category')
df['Genre'].dtypes
```

## What is the most Frequent genre of movies released on Netflix?

```python
df['Genre'].describe()
```

```python
sns.catplot( y = 'Genre' , data =df, kind = 'count' , order = df['Genre'].value_counts().index,
              hue='Genre',        
    palette='Set2',
    legend=False  )
plt.title('Genre Column Distribution',fontsize = 16)

plt.tight_layout()
plt.show()
```

```python
sns.catplot( y = 'Vote_Average' , data =df, kind = 'count' , order = df['Vote_Average'].value_counts().index,
              hue='Vote_Average',        
    palette='Set2',
    legend=False  )
plt.title('Vote Distribution',fontsize = 16)

plt.tight_layout()
plt.show()
```

## Moives Released per year

```python
plt.figure(figsize=(10,5))

sns.lineplot(
    x=df['Release_Date'].value_counts().sort_index().index,
    y=df['Release_Date'].value_counts().sort_index().values,
    marker='o'
)

plt.title('Movies Released Over Years', fontsize=14)
plt.xlabel('Year')
plt.ylabel('Number of Movies')
plt.tight_layout()
plt.grid(0.5)
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    df['Vote_Average'],
    bins=25,
    kde=True,
    color='#4CAF50',          # attractive green
    edgecolor='black',        # bar borders
    linewidth=2
)

plt.title('Distribution of Movie Ratings', fontsize=16, fontweight='bold')
plt.xlabel('Rating', fontsize=12)
plt.ylabel('Frequency', fontsize=12)

plt.grid(alpha=0)
plt.tight_layout()
plt.show()
```

## Top 10 most Popular Movies

```python
top_movies = df.sort_values(by='Popularity', ascending=False).head(15)

plt.figure(figsize=(10,6))

sns.barplot(
    x='Popularity',
    y='Title',
    data=top_movies,
    hue='Title',
    palette='coolwarm',
    legend=False
)

plt.title('Top 10 Most Popular Movies', fontsize=14)
plt.tight_layout()
plt.show()
```

## Least Popular Movies

```python
top_movies = df.sort_values(by='Popularity', ascending=True).head(10)

plt.figure(figsize=(10,6))

sns.barplot(
    x='Popularity',
    y='Title',
    data=top_movies,
    hue='Title',
    palette='viridis',
    legend=False
)

plt.title(' 10 Least Popular Movies', fontsize=14)
plt.tight_layout()
plt.show()
```
